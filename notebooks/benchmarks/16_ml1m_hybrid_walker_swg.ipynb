# ML-1M: original Sparse Walker + one residual SWG layer

Keeps the trained K=8 Walker frozen. Builds a fixed HNSW level-0 graph over the learned concept keys, starts navigation from the Walker's active concepts, and trains only the navigation query, sparse concept read projection, and residual gate.

Success target: validation NDCG@10 >= 0.145, ideally with most gain on 101–200 length histories.

```python
from google.colab import drive
drive.mount('/content/drive', force_remount=False)

import os, sys, shutil, subprocess, runpy
REPO='/content/Sparsewalker'
BRANCH='agent/walker-hybrid-swg'
if os.path.exists(REPO): shutil.rmtree(REPO)
subprocess.run(['git','clone','-q','-b',BRANCH,'https://github.com/hanialshater/Sparsewalker-.git',REPO],check=True)
subprocess.run([sys.executable,'-m','pip','install','-q','faiss-cpu'],check=True)
SRC=f'{REPO}/src'
EXP=f'{REPO}/experiments'
sys.path.insert(0,SRC)
sys.path.insert(0,EXP)
for name in list(sys.modules):
    if name=='sparsewalker' or name.startswith('sparsewalker.'):
        del sys.modules[name]
os.chdir(REPO)
sys.argv=['run_ml1m_hybrid_swg.py',
          '--seed','42',
          '--train-cap','120000',
          '--epochs','4',
          '--batch-size','1024',
          '--eval-batch-size','512',
          '--hnsw-m','8',
          '--ef-construction','160',
          '--hops','3',
          '--beam','8']
print('INPROCESS HYBRID RUN START', flush=True)
runpy.run_path(f'{EXP}/run_ml1m_hybrid_swg.py',run_name='__main__')
print('INPROCESS HYBRID RUN END', flush=True)
```
