# ML-1M temporal skips — fast launcher

This version skips the two expensive plain-Walker full-catalog evaluations that made the prior notebook look frozen. It also runs Python unbuffered, so epoch/eval progress appears immediately.

```python
from google.colab import drive
drive.mount('/content/drive', force_remount=False)

import os, sys, subprocess, shutil
REPO='/content/Sparsewalker'
BRANCH='agent/walker-temporal-skips'
if os.path.exists(REPO): shutil.rmtree(REPO)
subprocess.run(['git','clone','-q','-b',BRANCH,'https://github.com/hanialshater/Sparsewalker-.git',REPO], check=True)
SRC=f'{REPO}/src'
env=os.environ.copy()
env['PYTHONPATH']=SRC + (':' + env['PYTHONPATH'] if env.get('PYTHONPATH') else '')
env['PYTHONUNBUFFERED']='1'
cmd=[sys.executable,'-u',f'{REPO}/experiments/run_ml1m_temporal_skips_fast.py',
     '--seed','42','--max-epochs','12','--eval-every','2','--patience','6',
     '--batch-size','128','--eval-batch-size','1024']
print('RUNNING', ' '.join(cmd), flush=True)
subprocess.run(cmd,cwd=REPO,env=env,check=True)
```

```python
import pandas as pd
from pathlib import Path
p=Path('/content/drive/MyDrive/sparsewalker_temporal_skips_v2/ml1m/seed42/history.csv')
if p.exists():
    df=pd.read_csv(p)
    display(df[['epoch','loss','NDCG@10','HR@10','MRR@10','memory_share','seconds','positions_per_s']])
    print('best temporal val NDCG@10',df['NDCG@10'].max())
else:
    print('No history yet')
```
