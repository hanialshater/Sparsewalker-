# Sparse Walker temporal skips — fresh launcher

Single-cell launcher to avoid stale Colab state. It checks out the exact branch, imports directly from `/content/Sparsewalker/src`, runs a forward/backward smoke test, then launches the ML-1M temporal-skip experiment.


```python
from google.colab import drive
drive.mount('/content/drive', force_remount=False)

import os, sys, subprocess, shutil, pathlib, torch

REPO='/content/Sparsewalker'
BRANCH='agent/walker-temporal-skips'
if os.path.exists(REPO):
    shutil.rmtree(REPO)
subprocess.run(['git','clone','-q','-b',BRANCH,'https://github.com/hanialshater/Sparsewalker-.git',REPO], check=True)

# Import directly from source; no editable-install dependency.
SRC=f'{REPO}/src'
sys.path.insert(0, SRC)
os.environ['PYTHONPATH'] = SRC + (':' + os.environ['PYTHONPATH'] if os.environ.get('PYTHONPATH') else '')

import sparsewalker
print('sparsewalker from', sparsewalker.__file__)
from sparsewalker.models import SparseWalkerTemporalMemory

assert torch.cuda.is_available(), 'GPU runtime required'
print('GPU', torch.cuda.get_device_name(0), 'bf16', torch.cuda.is_bf16_supported())

m=SparseWalkerTemporalMemory(3706,200,d=64,layers=2,side=256,h=16,active=8,top_side=2,degree=4).cuda()
x=torch.randint(1,3707,(4,40),device='cuda')
with torch.autocast('cuda',dtype=torch.bfloat16):
    H=m(x)
    loss=H.float().square().mean()
loss.backward()
print('SMOKE OK', tuple(H.shape), 'loss', float(loss))
del m,x,H,loss
torch.cuda.empty_cache()

cmd=[sys.executable, f'{REPO}/experiments/run_ml1m_temporal_skips.py',
     '--seed','42','--max-epochs','20','--eval-every','2','--patience','8',
     '--batch-size','128','--eval-batch-size','1024']
print('RUNNING', ' '.join(cmd))
subprocess.run(cmd, cwd=REPO, env=os.environ.copy(), check=True)
```
